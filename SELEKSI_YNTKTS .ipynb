1. Urgensi dari perspektif grid management

Tentu saja mendesak karena seperti yang kita tahu grid management adalah proses mengelola dan mengoptimalkan penggunaan sumber daya dalam suatu sistem agar stabil dan aman. Dalam hal tersebut prediksi konsumsi listrik harian merupakan hal yang sangat penting karena dapat membantu mengoptimalkan dalam mengelola suatu sistem.
Demand-supply balance
Seperti yang kita ketahui. Listrik adalah komoditas yang unik karena tidak bisa disimpan dalam jumlah besar secara efisien. Artinya, listrik yang dihasilkan harus langsung digunakan pada saat itu juga. Jika produksinya lebih tinggi dari konsumsi, energi bisa terbuang. Sebaliknya, kalau konsumsi tiba-tiba melonjak dan tidak diimbangi oleh pasokan, bisa terjadi gangguan pada sistem. Nah, di sinilah pentingnya prediksi konsumsi harian—supaya operator seperti PLN bisa tahu berapa banyak listrik yang harus dibangkitkan dalam sehari, dan menyesuaikan operasi pembangkit secara tepat.  
Walaupun kita mengetahui bahwa pada zaman sekarang tiap pembangkit memiliki baterai skala besar yang mempunyai fungsi untuk menyimpan kelebihan produksi maupun menambah supply jika produksi saat itu tidak cukup hal ini berguna agar bisa menghandle lonjakan atau penurunan yang terjadi secara tiba-tiba jika diluar prediksi yang telah kita buat. Jika ini terjadi secra terus menerus maka lambat laun performa dari baterai akan berkurang atau bakan rusak, perlu dingat bahwa perbaikan atau penggantian baterai bukanlah hal yang murah. Maka peran dari prediksi konsumsi harian diharapkan dapat mengurangi beban kerja dari baterai dan secara tidak langsung dapat mengurangi presentase  penurunan performa dari baterai tersebut dan bahkan dapat menjaga baterai dari potensi kerusakan.

**Load Dispatching**

Dalam sistem kelistrikan, tidak semua pembangkit listrik dinyalakan secara bersamaan. Operator seperti PLN perlu menentukan pembangkit mana yang akan beroperasi terlebih dahulu, dan mana yang akan dinonaktifkan—semuanya disesuaikan dengan prediksi konsumsi harian.
Dengan mengetahui pola permintaan listrik, operator bisa mengatur urutan pembangkitan (load dispatch) berdasarkan efisiensi, biaya operasional, dan waktu respons pembangkit. Misalnya, pembangkit berbasis batu bara mungkin digunakan untuk beban dasar, sementara gas atau diesel dinyalakan saat beban puncak. Tanpa prediksi yang akurat, keputusan ini menjadi spekulatif dan berisiko menyebabkan pemborosan energi atau bahkan kekurangan daya.

**Monitoring Kinerja Jaringan dan Pembangkit**

Prediksi konsumsi listrik juga memainkan peran penting dalam pemantauan dan pengelolaan jaringan listrik. Dengan mengetahui sebaran beban secara harian, operator bisa menentukan margin cadangan yang dibutuhkan. Atau lebih tepatnya dapat menentukan selisih antara kapasitas pembangkit yang tersedia dengan beban maksimum yang diprediksi.
Lebih jauh lagi, prediksi membantu mengidentifikasi titik-titik dalam jaringan yang rawan beban berlebih (overload) atau underutilized. Ini memungkinkan operator melakukan antisipasi seperti redistribusi beban atau mengaktifkan pembangkit terdekat. Tanpa prediksi, tindakan semacam ini akan terlambat, dan risiko gangguan sistem (seperti blackout lokal) bisa meningkat.

I**ntegrasi Renewable Energy**

Pembangkit berbasis energi terbarukan seperti tenaga surya dan angin memiliki sifat intermiten, artinya produksinya naik turun tergantung kondisi cuaca dan waktu. Di sinilah prediksi konsumsi menjadi kunci agar operator bisa menentukan berapa banyak energi terbarukan yang bisa diserap jaringan tanpa mengorbankan kestabilan sistem.
Dengan proyeksi beban harian yang akurat, operator bisa mengatur kapan pembangkit renewable ditingkatkan, dan kapan perlu cadangan dari pembangkit konvensional atau baterai. Hal ini membantu mempercepat transisi energi bersih, sambil tetap menjaga keseimbangan antara supply dan demand secara real-time.


**Demand Response Management**

Dalam kondisi tertentu, seperti saat cuaca ekstrem atau lonjakan beban tak terduga, pasokan listrik bisa berada dalam kondisi kritis. Melalui prediksi konsumsi harian, operator bisa lebih awal mengidentifikasi potensi lonjakan beban tersebut.
Langkah yang bisa diambil adalah mengaktifkan demand response, yaitu mengirim sinyal atau insentif kepada konsumen besar (seperti industri) untuk menunda atau menurunkan konsumsi dalam jangka waktu tertentu. Ini seperti “mengendalikan” sisi permintaan untuk menyesuaikan dengan kapasitas pasokan yang tersedia. Jika dilakukan secara reaktif tanpa prediksi, responsnya bisa terlambat dan berisiko terjadi pemadaman.

2. Urgensi dari Perspektif Perencanaan Kapasitasas

**Investment Decision Making**

Dalam industri kelistrikan, keputusan investasi—baik untuk pembangunan pembangkit baru, jalur transmisi, maupun infrastruktur distribusi—merupakan keputusan strategis jangka panjang yang melibatkan dana triliunan rupiah dan waktu pengerjaan bertahun-tahun. Keputusan ini tidak bisa diambil sembarangan; dibutuhkan landasan data yang kuat dan dapat dipertanggungjawabkan. Di sinilah peran prediksi konsumsi harian dan tren beban jangka panjang menjadi sangat krusial.
Dengan menganalisis data puncak beban harian dari berbagai wilayah serta laju pertumbuhan konsumsi tahunan, operator seperti PLN dan pembuat kebijakan dapat menentukan waktu yang tepat dan lokasi yang paling membutuhkan pembangunan infrastruktur baru. Tanpa prediksi, proyek pembangunan bisa berujung pada dua skenario ekstrem:
•	Under-investment, yaitu pembangunan yang terlambat atau terlalu kecil skalanya, sehingga sistem kekurangan pasokan saat beban tumbuh, mengakibatkan pemadaman bergilir, keterbatasan ekspansi industri, dan kerugian ekonomi besar.
•	Over-investment, yaitu pembangunan yang terlalu cepat atau terlalu besar kapasitasnya, yang berujung pada infrastruktur "nganggur"—tidak termanfaatkan secara optimal dan hanya menjadi beban keuangan tanpa kontribusi nyata.
Prediksi konsumsi harian dan tren pertumbuhan adalah alat bantu vital untuk menghindari kedua skenario tersebut, serta memastikan bahwa setiap rupiah yang diinvestasikan benar-benar memberikan dampak positif terhadap keandalan dan efisiensi sistem tenaga listrik nasional.

**Generation Portfolio Optimization**

Sistem tenaga listrik yang sehat tidak hanya butuh kapasitas yang cukup, tetapi juga komposisi pembangkit yang optimal. Setiap jenis pembangkit memiliki karakteristik teknis dan biaya yang berbeda. Misalnya:
- Pembangkit baseload (seperti PLTU batubara atau gas combined-cycle) cocok untuk menyuplai beban dasar karena biaya operasinya murah dan efisiensinya tinggi, tetapi responsnya lambat jika terjadi lonjakan beban tiba-tiba.
- Pembangkit peaker (seperti PLTG gas atau diesel) dirancang untuk menyuplai beban puncak karena bisa dinyalakan cepat, namun biayanya jauh lebih mahal.
Dengan bantuan prediksi konsumsi harian, operator bisa menentukan kapan beban puncak akan terjadi dan berapa besar—sehingga peaker plants dapat disiapkan hanya saat dibutuhkan, bukan terus-menerus standby tanpa kejelasan. Sementara di sisi lain, pembangkit baseload bisa direncanakan untuk beroperasi secara penuh di luar jam sibuk, guna menurunkan total biaya produksi energi.
Tanpa prediksi, sistem bisa menjadi tidak efisien: peaker plants terlalu sering dinyalakan (biaya membengkak), atau baseload gagal memenuhi beban (mengakibatkan pemadaman). Maka, optimasi portofolio pembangkit bergantung sepenuhnya pada akurasi prediksi beban harian.

**Reserve Margin**

Dalam dunia ketenagalistrikan, margin cadangan (reserve margin) adalah salah satu parameter paling penting dalam menjamin keandalan sistem. Margin cadangan mengacu pada selisih antara kapasitas pembangkitan yang tersedia dan puncak beban yang diprediksi. Umumnya, standar internasional dan nasional menetapkan bahwa margin ini harus berada pada kisaran 15%–20% di atas puncak beban harian.
Namun, menjaga margin cadangan tidak semudah kelihatannya. Jika cadangan terlalu kecil, sistem akan kekurangan pasokan saat terjadi lonjakan mendadak atau gangguan pembangkit. Sebaliknya, jika terlalu besar, operator harus memelihara pembangkit cadangan yang jarang digunakan—yang artinya biaya tetap tinggi tanpa output nyata.
Di sinilah pentingnya prediksi beban harian yang akurat. Dengan prediksi yang tepat, operator dapat menetapkan margin cadangan secara presisi—cukup besar untuk menjamin keandalan, namun tidak berlebihan sehingga tidak membebani sistem dengan biaya tambahan yang tidak perlu.
Lebih dari itu, prediksi juga memungkinkan distribusi cadangan yang cerdas. Artinya, tidak hanya berapa besar cadangan yang dibutuhkan secara nasional, tetapi juga di mana cadangan itu harus tersedia—misalnya di wilayah dengan pertumbuhan konsumsi tinggi atau yang rawan gangguan sistem. Semua ini hanya bisa dilakukan jika operator memiliki dasar proyeksi beban yang kuat.

**Medium and Long Term Planning**

Prediksi konsumsi harian tidak hanya penting untuk operasional sehari-hari, tetapi juga menjadi fondasi perencanaan kapasitas jangka menengah hingga panjang dalam sistem tenaga listrik. Perusahaan seperti PLN perlu mengetahui bukan hanya kebutuhan energi hari ini, tetapi juga bagaimana kebutuhan itu akan berkembang dalam 5, 10, bahkan 20 tahun ke depan.
Dengan kombinasi data historis harian, tren mingguan, hingga pola musiman, operator dapat menyusun roadmap ekspansi kapasitas—yakni rencana pembangunan pembangkit baru, perluasan jaringan transmisi, serta pengembangan sistem penyimpanan energi. Keputusan seperti ini melibatkan investasi besar dengan jangka waktu realisasi yang panjang, sehingga tidak bisa didasarkan pada asumsi spekulatif.
Lebih jauh lagi, perencanaan jangka panjang tidak hanya didasarkan pada pola konsumsi saat ini, tetapi juga harus mempertimbangkan faktor eksternal yang dapat mengubah pola beban di masa depan, seperti:
•	Pertumbuhan ekonomi nasional, yang biasanya seiring dengan peningkatan konsumsi listrik industri dan rumah tangga.
•	Elektrifikasi transportasi, khususnya penetrasi kendaraan listrik (EV), yang akan menambah beban baru pada jaringan distribusi, terutama di malam hari saat pengisian baterai dilakukan.
•	Kebijakan transisi energi, seperti target bauran energi terbarukan nasional (misalnya 23% pada 2025), yang menuntut sistem lebih fleksibel dalam menyerap sumber energi bersifat intermiten.
Tanpa dukungan prediksi konsumsi jangka menengah dan panjang yang terstruktur, pengembangan sistem kelistrikan bisa berujung pada overbuilding (kelebihan kapasitas yang mubazir) atau sebaliknya, undercapacity yang menyebabkan krisis listrik di masa depan. Maka dari itu, prediksi bukan hanya alat bantu teknis, tapi instrumen strategis dalam pengambilan keputusan investasi sektor energi.

**Tariff Structure and Cost Efficiency**

Aspek lain yang sangat dipengaruhi oleh prediksi konsumsi harian adalah struktur tarif dan strategi efisiensi biaya sistem kelistrikan. Dengan memahami bagaimana beban listrik tersebar sepanjang hari, operator dapat merancang skema tarif yang lebih dinamis dan adil.
Salah satu contoh penerapan adalah time-of-use tariffs—yakni struktur tarif listrik yang berbeda berdasarkan waktu penggunaan. Konsumsi listrik pada jam sibuk (misalnya 18.00–22.00) dikenakan tarif lebih tinggi dibandingkan dengan jam sepi (misalnya dini hari). Tujuannya adalah menggeser sebagian konsumsi dari periode puncak ke periode off-peak, sehingga sistem kelistrikan tidak perlu menyalakan pembangkit cadangan (peaker plants) yang biayanya sangat mahal dan kurang efisien.
Tanpa prediksi beban harian, sulit bagi operator untuk menentukan jam-jam kritis yang memerlukan pengaturan konsumsi melalui mekanisme tarif. Sebaliknya, dengan pola konsumsi yang sudah dipetakan melalui prediksi, kebijakan tarif bisa disesuaikan secara akurat dan responsif terhadap kebutuhan jaringan.
Keuntungannya tidak hanya bagi sistem, tetapi juga bagi konsumen. Karena ketika sistem tidak perlu mengaktifkan pembangkit mahal, harga pokok produksi listrik menurun, dan pada akhirnya tarif listrik bagi masyarakat bisa lebih stabil dan kompetitif. Dengan kata lain, prediksi konsumsi yang akurat bukan hanya menjaga keseimbangan teknis sistem, tetapi juga menciptakan efisiensi ekonomi secara menyeluruh.

```python
import pandas as pd
from prophet import Prophet
from prophet.diagnostics import cross_validation, performance_metrics
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit
from sklearn.metrics import mean_squared_error
import numpy as np

# Dataframe untuk simpan hasil
results = []

# Loop per cluster
for cluster in data['cluster_id'].unique():
    print(f"\n=== Hybrid Prophet + RF | Cluster: {cluster} ===")

    cluster_data = data[data['cluster_id'] == cluster].copy()
    cluster_data.rename(columns={'date': 'ds', 'electricity_consumption': 'y'}, inplace=True)

    # Prophet tuning parameter grid
    param_grid = {
        'changepoint_prior_scale': [0.05, 0.1, 0.5],
        'seasonality_mode': ['additive', 'multiplicative']
    }

    best_rmse = float('inf')
    best_params = None

    for cp in param_grid['changepoint_prior_scale']:
        for sm in param_grid['seasonality_mode']:
            m = Prophet(changepoint_prior_scale=cp, seasonality_mode=sm, yearly_seasonality=True, weekly_seasonality=True)
            m.fit(cluster_data[['ds', 'y']])

            df_cv = cross_validation(m, initial='730 days', period='365 days', horizon='365 days', parallel="processes")
            df_p = performance_metrics(df_cv)
            prophet_rmse = df_p['rmse'].mean()

            if prophet_rmse < best_rmse:
                best_rmse = prophet_rmse
                best_params = {'changepoint_prior_scale': cp, 'seasonality_mode': sm}

    print(f"Best Prophet Params: {best_params}, RMSE: {best_rmse:.2f}")

    # Fit Prophet dengan param terbaik
    m_best = Prophet(**best_params, yearly_seasonality=True, weekly_seasonality=True)
    m_best.fit(cluster_data[['ds', 'y']])

    forecast = m_best.predict(cluster_data[['ds']])
    cluster_data['prophet_forecast'] = forecast['yhat'].values

    # Residual
    cluster_data['residual'] = cluster_data['y'] - cluster_data['prophet_forecast']

    # RF untuk residual
    feature_cols = ['temperature_2m_max', 'temperature_2m_min', 'apparent_temperature_max',
                    'apparent_temperature_min', 'sunshine_duration', 'daylight_duration',
                    'shortwave_radiation_sum', 'et0_fao_evapotranspiration','cluster_encoded']
    # feature_cols = num_features
    tscv = TimeSeriesSplit(n_splits=3)

    rf_model = RandomForestRegressor(random_state=42)

    param_grid_rf = {
        'n_estimators': [100, 200],
        'max_depth': [5, 10, 15],
        'min_samples_split': [2, 5]
    }

    grid_search = GridSearchCV(rf_model, param_grid_rf, cv=tscv, scoring='neg_root_mean_squared_error', n_jobs=-1)
    grid_search.fit(cluster_data[feature_cols], cluster_data['residual'])

    best_rf_rmse = -grid_search.best_score_
    print(f"Best RF Params: {grid_search.best_params_}, RF RMSE: {best_rf_rmse:.2f}")

    # Final Hybrid Prediksi
    pred_residual = grid_search.best_estimator_.predict(cluster_data[feature_cols])
    cluster_data['hybrid_pred'] = cluster_data['prophet_forecast'] + pred_residual

    hybrid_rmse = np.sqrt(mean_squared_error(cluster_data['y'], cluster_data['hybrid_pred']))
    print(f"Hybrid Prophet+RF Final RMSE: {hybrid_rmse:.2f}")

    # Simpan hasil ke dataframe
    results.append({
        'cluster_id': cluster,
        'prophet_rmse': best_rmse,
        'best_prophet_params': best_params,
        'rf_rmse': best_rf_rmse,
        'best_rf_params': grid_search.best_params_,
        'hybrid_rmse': hybrid_rmse
    })

# Buat dataframe hasil summary
results_df = pd.DataFrame(results)
print("\n=== Summary RMSE per Cluster ===")
print(results_df)
```
